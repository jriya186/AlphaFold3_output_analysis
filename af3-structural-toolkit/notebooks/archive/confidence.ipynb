```python
import os
import glob
import json
```

```python
def get_best_models_by_iptm_ptm(folder_path):
    """
    Find the best AlphaFold3 model(s) based on highest iPTM and pTM.
    If multiple models have the same iPTM and pTM, all are printed.
    """
    pattern = os.path.join(folder_path, "**", "*summary_confidences*.json")
    all_files = glob.glob(pattern, recursive=True)

    candidates = []

    # Collect valid models
    for file in all_files:
        try:
            with open(file, 'r') as f:
                data = json.load(f)

            iptm = data.get("iptm")
            ptm = data.get("ptm")

            if iptm is not None and ptm is not None:
                candidates.append({
                    "file": file,
                    "iptm": iptm,
                    "ptm": ptm
                })

        except Exception as e:
            print(f"Error reading {file}: {e}")

    if not candidates:
        print("No valid models with both iPTM and pTM found.")
        return

    # Step 1: Get highest iPTM
    max_iptm = max(c["iptm"] for c in candidates)
    top_iptm_models = [c for c in candidates if c["iptm"] == max_iptm]

    # Step 2: Among those, get highest pTM
    max_ptm = max(c["ptm"] for c in top_iptm_models)
    best_models = [c for c in top_iptm_models if c["ptm"] == max_ptm]

    # Report
    print(f"Best iPTM: {max_iptm}")
    print(f"Best pTM: {max_ptm}\n")
    print("Best model(s):")
    for model in best_models:
        print(f"File: {model['file']}")
        print(f"  iPTM: {model['iptm']}")
        print(f"  pTM: {model['ptm']}\n")
```

```python
# Replace this with your actual output folder path
folder = "/path/fold_apoer2ex3_6r562ca"

get_best_models_by_iptm_ptm(folder)
```
